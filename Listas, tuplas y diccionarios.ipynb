```python
#28
clientes = [
 {"nombre": "Ana", "riesgo": "bajo"},
 {"nombre": "Luis", "riesgo": "alto"},
 {"nombre": "Marta", "riesgo": "bajo"}
]
grupo = {}

for c in clientes:
 riesgo = c["riesgo"]
 if riesgo not in grupo:
  grupo[riesgo] = []
 grupo[riesgo].append(c["nombre"])
print(grupo)
```

```python
# 30
transacciones = [
 {"cliente": "Ana", "tipo": "Credito", "monto": 120000},
 {"cliente": "Luis", "tipo": "Debito", "monto": 85000},
 {"cliente": "Marta", "tipo": "Credito", "monto": 210000}
]
total_creditos = 0
clientes = []
for t in transacciones:
    if t["tipo"] == "Credito" and t["monto"] >= 100000:
        total_creditos += t["monto"]
        clientes.append(t["cliente"])
print(total_creditos)
print(clientes)
```

**RTA/:**
las Listas guardan todas las transacciones, los Diccionarios guardan los datos de cada cliente.
for: revisa cada transacción.
if: selecciona las que son créditos de $100.000 o más
El acumulador: suma el valor de esos créditos.
Lista clientes: guarda los nombres que cumplen la condición.
